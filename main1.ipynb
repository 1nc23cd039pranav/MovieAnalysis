```python
#function to categorize the 'Vote_Average' column into 4 categories based on the quartiles
def categorize_col(df, col,labels):
    edges=[df[col].describe()['min'],
           df[col].describe()['25%'],
           df[col].describe()['50%'],
           df[col].describe()['75%'],
           df[col].describe()['max']]
    df[col] = pd.cut(df[col], edges, labels = labels, duplicates='drop')
    return df
```

```python
labels = ['not popular','below average','average', 'popular']
categorize_col(df,'Vote_Average',labels)
df['Vote_Average'].unique()
```

# Analysis Visualizations

```python
# Reload data for analysis (keeping all original columns)
df_analysis = pd.read_csv('movie.csv', lineterminator='\n')
df_analysis['Release_Date'] = pd.to_datetime(df_analysis['Release_Date'])
```

## 1. Relationship Between Popularity and Quality

```python
plt.figure(figsize=(12, 6))
plt.scatter(df_analysis['Popularity'], df_analysis['Vote_Average'], alpha=0.6, s=100, color='#42a5f5')
plt.xlabel('Popularity Score', fontsize=12)
plt.ylabel('Vote Average (Rating)', fontsize=12)
plt.title('Relationship Between Popularity and Quality (Rating)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)

# Add correlation coefficient
correlation = df_analysis['Popularity'].corr(df_analysis['Vote_Average'])
plt.text(0.02, 0.98, f'Correlation: {correlation:.3f}', 
         transform=plt.gca().transAxes, fontsize=11, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
plt.tight_layout()
plt.show()

print(f"Correlation between Popularity and Vote Average: {correlation:.3f}")
```

## 2. Genre Performance Analysis

```python
# Create a dataframe with exploded genres for analysis
df_genre = df_analysis.copy()
df_genre['Genre'] = df_genre['Genre'].str.split(', ')
df_genre = df_genre.explode('Genre')

# Calculate average metrics by genre
genre_stats = df_genre.groupby('Genre').agg({
    'Vote_Average': 'mean',
    'Popularity': 'mean',
    'Vote_Count': 'mean',
    'Title': 'count'
}).round(2)
genre_stats.columns = ['Avg Rating', 'Avg Popularity', 'Avg Votes', 'Count']
genre_stats = genre_stats.sort_values('Avg Rating', ascending=False)

# Create subplots for genre analysis
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Plot 1: Average Rating by Genre
genre_stats['Avg Rating'].plot(kind='barh', ax=axes[0, 0], color='#42a5f5')
axes[0, 0].set_title('Average Rating by Genre', fontweight='bold')
axes[0, 0].set_xlabel('Average Vote')

# Plot 2: Average Popularity by Genre
genre_stats['Avg Popularity'].plot(kind='barh', ax=axes[0, 1], color='#66bb6a')
axes[0, 1].set_title('Average Popularity by Genre', fontweight='bold')
axes[0, 1].set_xlabel('Average Popularity')

# Plot 3: Movie Count by Genre
genre_stats['Count'].plot(kind='barh', ax=axes[1, 0], color='#ffa726')
axes[1, 0].set_title('Number of Movies by Genre', fontweight='bold')
axes[1, 0].set_xlabel('Count')

# Plot 4: Average Vote Count by Genre
genre_stats['Avg Votes'].plot(kind='barh', ax=axes[1, 1], color='#ab47bc')
axes[1, 1].set_title('Average Vote Count by Genre', fontweight='bold')
axes[1, 1].set_xlabel('Average Vote Count')

plt.tight_layout()
plt.show()

print("\nGenre Performance Statistics:")
print(genre_stats)
```

## 3. Temporal Trends - Release Date Impact

```python
# Extract year and month for temporal analysis
df_analysis['Release_Year'] = df_analysis['Release_Date'].dt.year
df_analysis['Release_Month'] = df_analysis['Release_Date'].dt.month
df_analysis['Release_Month_Name'] = df_analysis['Release_Date'].dt.strftime('%B')

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Plot 1: Average Rating by Release Year
yearly_stats = df_analysis.groupby('Release_Year')['Vote_Average'].agg(['mean', 'count'])
axes[0].plot(yearly_stats.index, yearly_stats['mean'], marker='o', linewidth=2, markersize=8, color='#42a5f5')
axes[0].fill_between(yearly_stats.index, yearly_stats['mean'], alpha=0.3, color='#42a5f5')
axes[0].set_title('Average Rating by Release Year', fontweight='bold', fontsize=12)
axes[0].set_xlabel('Release Year')
axes[0].set_ylabel('Average Rating')
axes[0].grid(True, alpha=0.3)

# Plot 2: Average Rating by Release Month
monthly_stats = df_analysis.groupby('Release_Month_Name')['Vote_Average'].mean()
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 
               'September', 'October', 'November', 'December']
monthly_stats = monthly_stats.reindex([m for m in month_order if m in monthly_stats.index])
monthly_stats.plot(kind='bar', ax=axes[1], color='#66bb6a')
axes[1].set_title('Average Rating by Release Month', fontweight='bold', fontsize=12)
axes[1].set_xlabel('Release Month')
axes[1].set_ylabel('Average Rating')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

print("\nAverage Rating by Year:")
print(yearly_stats)
```

## 4. Impact of Language on Movie Success

```python
# Get top languages
language_stats = df_analysis.groupby('Original_Language').agg({
    'Vote_Average': ['mean', 'std'],
    'Popularity': ['mean', 'std'],
    'Vote_Count': ['mean', 'std'],
    'Title': 'count'
}).round(2)
language_stats.columns = ['Avg Rating', 'Std Rating', 'Avg Pop', 'Std Pop', 'Avg Votes', 'Std Votes', 'Count']
language_stats = language_stats[language_stats['Count'] >= 2]  # Filter for languages with at least 2 movies
language_stats = language_stats.sort_values('Count', ascending=False).head(10)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Plot 1: Average Rating by Top Languages
language_stats['Avg Rating'].plot(kind='barh', ax=axes[0], color='#42a5f5', xerr=language_stats['Std Rating'])
axes[0].set_title('Top 10 Languages - Average Rating', fontweight='bold')
axes[0].set_xlabel('Average Rating')

# Plot 2: Average Popularity by Top Languages
language_stats['Avg Pop'].plot(kind='barh', ax=axes[1], color='#66bb6a', xerr=language_stats['Std Pop'])
axes[1].set_title('Top 10 Languages - Average Popularity', fontweight='bold')
axes[1].set_xlabel('Average Popularity')

# Plot 3: Movie Count by Top Languages
language_stats['Count'].plot(kind='barh', ax=axes[2], color='#ffa726')
axes[2].set_title('Top 10 Languages - Movie Count', fontweight='bold')
axes[2].set_xlabel('Number of Movies')

plt.tight_layout()
plt.show()

print("\nTop Languages Performance Statistics:")
print(language_stats)
```

## 5. Vote Engagement Analysis

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Plot 1: Vote Count vs Vote Average (Scatter)
axes[0, 0].scatter(df_analysis['Vote_Count'], df_analysis['Vote_Average'], 
                   alpha=0.6, s=100, c=df_analysis['Popularity'], cmap='viridis')
axes[0, 0].set_title('Vote Count vs Vote Average', fontweight='bold')
axes[0, 0].set_xlabel('Vote Count (Engagement)')
axes[0, 0].set_ylabel('Vote Average (Rating)')
axes[0, 0].grid(True, alpha=0.3)

# Calculate correlation
vote_correlation = df_analysis['Vote_Count'].corr(df_analysis['Vote_Average'])
axes[0, 0].text(0.02, 0.98, f'Correlation: {vote_correlation:.3f}', 
               transform=axes[0, 0].transAxes, fontsize=10, verticalalignment='top',
               bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# Plot 2: Distribution of Vote Count
axes[0, 1].hist(df_analysis['Vote_Count'], bins=30, color='#42a5f5', edgecolor='black', alpha=0.7)
axes[0, 1].set_title('Distribution of Vote Counts', fontweight='bold')
axes[0, 1].set_xlabel('Vote Count')
axes[0, 1].set_ylabel('Frequency')

# Plot 3: Vote Average Distribution
axes[1, 0].hist(df_analysis['Vote_Average'], bins=20, color='#66bb6a', edgecolor='black', alpha=0.7)
axes[1, 0].set_title('Distribution of Vote Average (Ratings)', fontweight='bold')
axes[1, 0].set_xlabel('Vote Average')
axes[1, 0].set_ylabel('Frequency')

# Plot 4: Popularity vs Vote Count
axes[1, 1].scatter(df_analysis['Vote_Count'], df_analysis['Popularity'], 
                   alpha=0.6, s=100, color='#ab47bc')
axes[1, 1].set_title('Popularity vs Vote Count', fontweight='bold')
axes[1, 1].set_xlabel('Vote Count (Engagement)')
axes[1, 1].set_ylabel('Popularity Score')
axes[1, 1].grid(True, alpha=0.3)

# Calculate correlation
pop_vote_correlation = df_analysis['Vote_Count'].corr(df_analysis['Popularity'])
axes[1, 1].text(0.02, 0.98, f'Correlation: {pop_vote_correlation:.3f}', 
               transform=axes[1, 1].transAxes, fontsize=10, verticalalignment='top',
               bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()

print(f"\nVote Engagement Analysis:")
print(f"Correlation (Vote Count vs Rating): {vote_correlation:.3f}")
print(f"Correlation (Vote Count vs Popularity): {pop_vote_correlation:.3f}")
print(f"\nBasic Statistics:")
print(f"Average Vote Count: {df_analysis['Vote_Count'].mean():.0f}")
print(f"Median Vote Count: {df_analysis['Vote_Count'].median():.0f}")
print(f"Average Vote Average: {df_analysis['Vote_Average'].mean():.2f}")
print(f"Median Vote Average: {df_analysis['Vote_Average'].median():.2f}")
```
